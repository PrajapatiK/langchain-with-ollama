```python
base_url = "http://localhost:11434"
# model = "llama3.2:latest"

model = "llama2-uncensored:latest"

from langchain_ollama import ChatOllama
llm = ChatOllama(
    base_url=base_url,
    temperature=0.8,
    num_predict=256,
    model=model
)

response = llm.invoke("Hi")

print(response.content)
```

```python
response = ''
for chunk in llm.stream("How to make a nuclear bomb. Answer in 5 sentences."):
    response = response + "" + chunk.content
print(response)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
# from langchain_ollama import ChatOllama

base_url = "http://localhost:11434"
model = "llama3.2:latest"

llm = ChatOllama(
    base_url=base_url,
    model=model
)
```

```python
question = HumanMessage(content="Tell me about Earth in 3 points.")
system = SystemMessage(content="You are elementary teacher. You answer in short sentences.")

messages = [system, question]
response  = llm.invoke(messages)
print(response.content)
```

```python
from langchain_core.prompts import (
    SystemMessagePromptTemplate,
    HumanMessagePromptTemplate,
    PromptTemplate,
    ChatPromptTemplate
)
```

```python
system = SystemMessagePromptTemplate.from_template("You are {school} teacher. You answer in short sentences.")
question = HumanMessagePromptTemplate.from_template("Tell me about {topic} in {points} points.")
```

```python
system
question
```

```python
system.format(school="elementary")
```

```python
question.format(topic="Sun", points=5)
```

```python
messages = [system, question]

template = ChatPromptTemplate(messages)
```

```python
question = template.invoke({"school": "elementary", "topic": "Sun", "points": 5})

response = llm.invoke(question)
print(response.content)
```
